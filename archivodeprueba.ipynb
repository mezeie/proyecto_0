```python
def separar_viento(campo_viento):
    campo_viento = campo_viento.strip()

    if campo_viento == "" or campo_viento.lower() == "no se calcula":
        return (None, None)


    if campo_viento.lower() == "calma":
        return ("Calma", 0)

    palabras = campo_viento.split()


    if palabras and palabras[-1].isdigit():
        velocidad = int(palabras[-1])
        direccion = " ".join(palabras[:-1])
        return (direccion, velocidad)


    return (campo_viento, None)
```

```python
def leer_observaciones(ruta):
    observaciones = {}
    lineas_invalidas = 0

    try:
        f = open(ruta, "r", encoding="latin-1")

        for linea in f:
            linea = linea.strip()


            if not linea:
                continue

            if linea.endswith("/"):
                linea = linea[:-1].strip()

            campos = linea.split(";")

            if len(campos) != 10:
                lineas_invalidas += 1
                continue


            ciudad = campos[0].strip()


            fecha_texto = campos[1].strip()
            partes_fecha = fecha_texto.split("-")

            if len(partes_fecha) == 3:
                try:
                    d = int(partes_fecha[0])
                    m = MESES.get(partes_fecha[1].lower(), 1)
                    a = int(partes_fecha[2])
                    fecha_parseada = datetime(a, m, d).date()
                except ValueError:
                    fecha_parseada = None
            else:
                fecha_parseada = None


            temp_texto = campos[5].strip()
            if temp_texto.lower() == "no se calcula":
                temp = None
            else:
                try:
                    temp = float(temp_texto)
                except ValueError:
                    temp = None


            st_texto = campos[6].strip()
            if st_texto.lower() == "no se calcula":
                sensacion = None
            else:
                try:
                    sensacion = float(st_texto)
                except ValueError:
                    sensacion = None

            dir_viento, vel_viento = separar_viento(campos[8])


            observaciones[ciudad] = {
                "fecha": fecha_parseada,
                "hora": campos[2].strip(),
                "condicion": campos[3].strip(),
                "visibilidad": campos[4].strip(),
                "temperatura": temp,
                "sensacion_termica": sensacion,
                "humedad": campos[7].strip(),
                "direccion_viento": dir_viento,
                "velocidad_viento": vel_viento,
                "presion": campos[9].strip()
            }


        f.close()

    except FileNotFoundError:
        print("error no esta el archivo")
        return {}, 0


    return observaciones, lineas_invalidas
```

```python
def cantidad_ciudades_completas(observaciones):
    c = 0

    for datos in observaciones.values():
        if None not in datos.values():
            c += 1


    return c


print(cantidad_ciudades_completas(obs))
```

```python
def datos_faltantes_por_campo(observaciones):
    faltantes = {}


    for ciudad, datos in observaciones.items():
        for campo, valor in datos.items():

            if valor is None:
                if campo not in faltantes:
                    faltantes[campo] = {"cantidad": 0, "estaciones": []}

                faltantes[campo]["cantidad"] += 1
                faltantes[campo]["estaciones"].append(ciudad)


    return faltantes
```

```python
def top_n_ciudades(observaciones, campo, n, descendente=True):
    ciudades_validas = []


    for ciudad, datos in observaciones.items():
        if datos.get(campo) is not None:
            ciudades_validas.append((datos[campo], ciudad))


    ciudades_validas.sort()


    if descendente:
        ciudades_validas = ciudades_validas[::-1]


    seleccionadas = ciudades_validas[:n]

    resultado = []

    for valor, ciudad in seleccionadas:
        resultado.append((ciudad, valor))

    return resultado
```

```python
def mostrar_resumen(observaciones, lineas_invalidas):
    if not observaciones:
        print("No hay nada cargado")
        return

    lineas = []

    lineas.append("================ RESUMEN METEOROLOGICO ================")
    lineas.append(f"Total de ciudades leidas: {cantidad_ciudades(observaciones)}")
    lineas.append(f"Ciudades con datos completos: {cantidad_ciudades_completas(observaciones)}")
    lineas.append(f"Lineas mal formadas o invalidas: {lineas_invalidas}\n")


    faltantes = datos_faltantes_por_campo(observaciones)
    lineas.append("DATOS FALTANTES POR CAMPO")

    if not faltantes:
        lineas.append("No se registraron datos faltantes.")
    else:
        for campo, info in faltantes.items():
            estaciones = info["estaciones"]
            cant = info["cantidad"]

            if cant > 5:
                estaciones_str = ", ".join(estaciones[:5]) + f" y {cant - 5} estaciones mas"
            else:
                estaciones_str = ", ".join(estaciones)

            lineas.append(f"- {campo}: {cant} faltante(s) en {estaciones_str}")


    lineas.append("")


    temp_max = top_n_ciudades(observaciones, "temperatura", 1, descendente=True)
    temp_min = top_n_ciudades(observaciones, "temperatura", 1, descendente=False)
    viento_max = top_n_ciudades(observaciones, "velocidad_viento", 1, descendente=True)
    viento_min = top_n_ciudades(observaciones, "velocidad_viento", 1, descendente=False)


    lineas.append("VALORES EXTREMOS")
    if temp_max:
        lineas.append(f"Temperatura mas alta: {temp_max[0][0]} con {temp_max[0][1]} °C")
    if temp_min:
        lineas.append(f"Temperatura mas baja: {temp_min[0][0]} con {temp_min[0][1]} °C")
    if viento_max:
        lineas.append(f"Viento mas fuerte: {viento_max[0][0]} a {viento_max[0][1]} km/h")
    if viento_min:
        lineas.append(f"Viento mas suave: {viento_min[0][0]} a {viento_min[0][1]} km/h")


    lineas.append("")


    lineas.append("CIUDADES MAS CALIDAS")
    for ciudad, temp in top_n_ciudades(observaciones, "temperatura", 5, descendente=True):
        lineas.append(f"  {ciudad}: {temp} °C")


    lineas.append("\nCIUDADES MAS FRIAS")
    for ciudad, temp in top_n_ciudades(observaciones, "temperatura", 5, descendente=False):
        lineas.append(f"  {ciudad}: {temp} °C")


    lineas.append("\nCIUDADES CON MAS VIENTO")
    for ciudad, vel in top_n_ciudades(observaciones, "velocidad_viento", 5, descendente=True):
        lineas.append(f"  {ciudad}: {vel} km/h")


    lineas.append("\nCIUDADES CON MENOS VIENTO")
    for ciudad, vel in top_n_ciudades(observaciones, "velocidad_viento", 5, descendente=False):
        lineas.append(f"  {ciudad}: {vel} km/h")


    lineas.append("=======================================================")
```

```python
mostrar_resumen(obs, inv)
```
